```python
import requests
import csv
```

```python
# --- Traer datos de la API ---
URL = 'https://www.datos.gov.co/resource/mqpd-2jhs.json'
parametros = {'$limit': 600000}

respuesta = requests.get(URL, params=parametros)
datos = respuesta.json()

print(f'Registros obtenidos: {len(datos)}')
print(f'Campos por registro: {len(datos[0].keys())}')
print(f'Primeras claves: {list(datos[0].keys())[:5]}')
```

```python
# --- Seleccionar columnas relevantes ---
COLUMNAS = [
    'ins_sede_nombre', 'programa', 'dep_nac', 'sexo', 'edad',
    'estrato', 'tipo_ins', 'paes', 'ptotal', 'admitido',
    'area_cine', 'discapacidad', 'nacionalidad', 'cods_nac', 
    'year', 'semestre'
]

datos_filtrados = []
for registro in datos:
    fila = {}
    for col in COLUMNAS:
        fila[col] = registro.get(col, '')
    datos_filtrados.append(fila)

print(f'\nColumnas seleccionadas: {len(COLUMNAS)}')
print(f'Registros filtrados: {len(datos_filtrados)}')
```

```python
# --- Escribir CSV ---
ARCHIVO = 'aspirantes_unal.csv'

with open(ARCHIVO, 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=COLUMNAS)
    writer.writeheader()
    writer.writerows(datos_filtrados)

print(f'\nArchivo {ARCHIVO} creado.')
```

```python
with open(ARCHIVO, 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    filas = list(reader)
    print(f'Verificación: {len(filas)} filas, {len(filas[0].keys())} columnas')
    print(f'Primera fila: {dict(list(filas[0].items())[:4])}...')
```
